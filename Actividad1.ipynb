#### Funciones y Manejo de Errores
Crea código reutilizable y a prueba de errores de usuario siguiendo estos parámetros.

1. Crea una función llamada calcular_ingreso que reciba dos parámetros: precio y cantidad.
Dentro de la función, utiliza un bloque try-except. Intenta multiplicar el precio por la cantidad.
2. Si ocurre un error (por ejemplo, si en lugar de un número llega un texto), el except debe atraparlo, imprimir "Error: Los valores deben ser numéricos" y retornar 0.
3. Prueba tu función enviando valores correctos y luego enviando un texto para ver cómo se maneja el error.

```python
def calcular_ingreso(precio, cantidad):
    try:
        precio = float(precio)
        cantidad = float(cantidad)

        return precio * cantidad

    except ValueError:
        print("Error: Los valores deben ser numéricos")
        return 0

    except Exception as e:
        print(f"Error inesperado: {e}")
        return 0
# Ejemplo de uso
precio_producto = 45000
cantidad_productos = 2
ingreso_total = calcular_ingreso(precio_producto, cantidad_productos)
print(f"El ingreso total por la venta es: ${ingreso_total: }")

#poniendo a prueba la función con un valor no numérico
precio_producto = "cuarenta y cinco mil"    
cantidad_productos = 2
ingreso_total = calcular_ingreso(precio_producto, cantidad_productos)
print(f"El ingreso total por la venta es: ${ingreso_total: }")

#Error inesperado
precio_producto = 45000
cantidad_productos = None
ingreso_total = calcular_ingreso(precio_producto, cantidad_productos)
print(f"El ingreso total por la venta es: ${ingreso_total: }")


```
